# Capítulo 2: Cadenas de Markov y Distribución Estacionaria

En este cuaderno analizamos la dinámica temporal estocástica de una cinta transportadora del almacén mediante una **Cadena de Markov de Tiempo Discreto**.

### Estados de la Máquina:
1. **Nuevo (N)**
2. **Desgastado (D)**
3. **Averiado (A)**

### Matriz de Transición $P$:
$$P = \begin{pmatrix} 0.70 & 0.20 & 0.10 \\ 0.10 & 0.60 & 0.30 \\ 0.10 & 0.00 & 0.90 \end{pmatrix}$$


```python
import numpy as np

# 1. Definición de la matriz estocástica de transición P
P = np.array([
    [0.70, 0.20, 0.10],  # Fila 1: Nuevo (N)
    [0.10, 0.60, 0.30],  # Fila 2: Desgastado (D)
    [0.10, 0.00, 0.90]   # Fila 3: Averiado (A)
])

# 2. Proyección temporal a corto plazo (t=1 y t=2) partiendo de pieza Nueva
mu_0 = np.array([1.0, 0.0, 0.0])
mu_1 = mu_0 @ P
mu_2 = mu_0 @ np.linalg.matrix_power(P, 2)

print("=== PROYECCIÓN A CORTO PLAZO ===")
print("Distribución en t=1 (mu_1):", np.round(mu_1, 3))
print("Distribución en t=2 (mu_2):", np.round(mu_2, 3))
print(f"Probabilidad de avería en t=2: {mu_2[2]*100:.1f}%
")
```

## Convergencia por Multiplicaciones Sucesivas ($P^k$)
A medida que $k \to \infty$, las filas de $P^k$ convergen a la misma distribución estacionaria $\pi^*$ independiente del estado inicial.


```python
print("=== CONVERGENCIA POR MULTIPLICACIÓN MATRICIAL SUCESIVA ===")
for k in [1, 2, 5, 10, 50]:
    Pk = np.linalg.matrix_power(P, k)
    print(f"Matriz P^{k}:")
    print(np.round(Pk, 4))
    print()
```

## Resolución Algebraica Exacta mediante Autovectores
Calculamos la distribución estacionaria resolviendo $\pi^* P = \pi^*$ (autovalor $\lambda = 1$).


```python
# 4. Resolución algebraica exacta mediante autovectores (pi * P = pi)
evals, evecs = np.linalg.eig(P.T)
idx = np.argmin(np.abs(evals - 1.0))
pi_star = np.real(evecs[:, idx])
pi_star = pi_star / np.sum(pi_star)  # Normalizamos para que sume 1.0

estados = ["Nuevo (N)", "Desgastado (D)", "Averiado (A)"]
print("=== DISTRIBUCIÓN ESTACIONARIA ASINTÓTICA (pi*) ===")
for st, prob in zip(estados, pi_star):
    print(f"  {st:16s}: {prob*100:6.2f}% del tiempo total")
```
