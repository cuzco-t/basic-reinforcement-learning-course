# Capítulo 0: Fundamentos de Probabilidad y Ley de los Grandes Números

En este cuaderno interactivo validamos empíricamente el concepto de **Valor Esperado** y la **Ley de los Grandes Números** aplicados al caso de estudio de **OptiBot** en el almacén inteligente.

### Escenario:
OptiBot realiza despachos de paquetes con tres posibles resultados económicos:
1. **Entrega a tiempo:** Ganancia de $+10.0$ USD ($P = 0.26$)
2. **Entrega estándar:** Ganancia de $+5.0$ USD ($P = 0.50$)
3. **Paquete dañado / retraso:** Pérdida de $-2.0$ USD ($P = 0.24$)

El valor esperado teórico se calcula como:
$$E[X] = \sum_{x} x \cdot P(X = x)$$


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definición de los resultados posibles y sus probabilidades
ganancias = np.array([10.0, 5.0, -2.0])
probabilidades = np.array([0.26, 0.50, 0.24])

# 2. Cómputo del valor esperado teórico
E_teorico = np.sum(ganancias * probabilidades)
print(f"Valor Esperado Teórico: ${E_teorico:.2f}")
```

## Simulación de Monte Carlo: Ley de los Grandes Números
Simulamos 50,000 despachos aleatorios para comprobar que la media muestral converge al valor esperado teórico.


```python
# 3. Simulación de 50,000 despachos realizados por OptiBot
np.random.seed(42)
n_entregas = 50000
despachos = np.random.choice(ganancias, size=n_entregas, p=probabilidades)

# 4. Promedio real obtenido a partir de la simulación
promedio_observado = np.mean(despachos)
print(f"Promedio Real Observado ({n_entregas:,} entregas): ${promedio_observado:.2f}")

# 5. Comparación entre la predicción teórica y el resultado práctico
diferencia = abs(E_teorico - promedio_observado)
print(f"Diferencia absoluta: ${diferencia:.4f}")
```

## Visualización de la Convergencia
Graficamos la evolución del promedio acumulado a medida que aumenta el número de entregas.


```python
# 6. Gráfica de convergencia del promedio acumulado
promedio_acumulado = np.cumsum(despachos) / np.arange(1, n_entregas + 1)

plt.figure(figsize=(10, 5))
plt.plot(promedio_acumulado[:5000], label='Promedio Acumulado Muestral', color='#1f77b4', lw=1.5)
plt.axhline(E_teorico, color='#d62728', linestyle='--', label=f'Valor Esperado Teórico (${E_teorico:.2f})', lw=2)
plt.title('Ley de los Grandes Números: Convergencia de OptiBot')
plt.xlabel('Número de Entregas')
plt.ylabel('Ganancia Promedio por Entrega ($)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
